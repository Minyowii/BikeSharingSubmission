## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

### Cleaning Data

#### Data Cleaning and Quality Improvement

```python
# Membuat salinan dataset agar data asli tetap tersedia
df_clean = df.copy()

# Mengubah kolom tanggal menjadi tipe datetime
df_clean["dteday"] = pd.to_datetime(
    df_clean["dteday"],
    errors="coerce"
)

# Memeriksa kembali kualitas data
print("Missing values sebelum cleaning:")
display(df_clean.isnull().sum())

print("Jumlah baris duplikat sebelum cleaning:",
      df_clean.duplicated().sum())

# Menghapus baris duplikat jika ditemukan
df_clean = df_clean.drop_duplicates()

# Menghapus baris dengan tanggal tidak valid jika ditemukan
df_clean = df_clean.dropna(subset=["dteday"])

# Mengurutkan data berdasarkan tanggal
df_clean = df_clean.sort_values("dteday").reset_index(drop=True)

# Memeriksa hasil cleaning
print("Ukuran dataset sebelum cleaning:", df.shape)
print("Ukuran dataset setelah cleaning:", df_clean.shape)
print("Total missing values setelah cleaning:",
      df_clean.isnull().sum().sum())
print("Total baris duplikat setelah cleaning:",
      df_clean.duplicated().sum())

display(df_clean.head())
```

**Insight:** (Opsional)
- Kolom dteday dikonversi menjadi tipe data tanggal agar dapat digunakan untuk analisis berdasarkan bulan dan tahun.
- Data duplikat dan tanggal yang tidak valid dihapus apabila ditemukan. Hasil pemeriksaan setelah cleaning digunakan untuk memastikan dataset siap digunakan dalam proses eksplorasi dan visualisasi.

## Exploratory Data Analysis (EDA)

### Explore Exploring Bike Rental Patterns by Time and Weather

```python
# Membuat kolom tahun dan nama bulan untuk analisis
df_clean["year"] = df_clean["dteday"].dt.year
df_clean["month_num"] = df_clean["dteday"].dt.month

# Mengubah kode musim menjadi nama musim
season_mapping = {
    1: "Winter",
    2: "Spring",
    3: "Summer",
    4: "Fall"
}
df_clean["season_name"] = df_clean["season"].map(season_mapping)

# Mengubah kode cuaca menjadi deskripsi
weather_mapping = {
    1: "Clear / Few Clouds",
    2: "Mist / Cloudy",
    3: "Light Rain / Snow",
    4: "Heavy Rain / Snow"
}
df_clean["weather_name"] = df_clean["weathersit"].map(weather_mapping)

# Statistik deskriptif jumlah penyewaan sepeda
display(df_clean[["casual", "registered", "cnt"]].describe().T)

# Rata-rata penyewaan harian berdasarkan bulan
monthly_rentals = (
    df_clean.groupby("month_num")["cnt"]
    .agg(["mean", "median", "sum", "count"])
    .reindex(range(1, 13))
)

monthly_rentals.index = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]
monthly_rentals.index.name = "Month"
display(monthly_rentals.round(2))

# Rata-rata penyewaan harian berdasarkan musim
seasonal_rentals = (
    df_clean.groupby("season_name")["cnt"]
    .agg(["mean", "median", "sum", "count"])
    .reindex(["Winter", "Spring", "Summer", "Fall"])
)
display(seasonal_rentals.round(2))

# Rata-rata penyewaan harian berdasarkan cuaca
weather_rentals = (
    df_clean.groupby("weather_name")["cnt"]
    .agg(["mean", "median", "sum", "count"])
    .sort_values("mean", ascending=False)
)
display(weather_rentals.round(2))
```

**Insight:** (Opsional)
- Analisis eksploratif dilakukan untuk memahami distribusi jumlah penyewaan sepeda, baik dari pengguna kasual (casual) maupun pengguna terdaftar (registered), serta melihat perbedaan jumlah penyewaan berdasarkan bulan, musim, dan kondisi cuaca.



- Pengelompokan berdasarkan waktu dan cuaca membantu mengidentifikasi pola permintaan penyewaan sepeda sehingga dapat digunakan untuk menjawab pertanyaan bisnis pada tahap visualisasi.

## Visualization & Explanatory Analysis

### Bike Rental Patterns by Month and Season

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Rata-rata penyewaan berdasarkan bulan
monthly_avg = (
    df_clean.groupby(df_clean["dteday"].dt.month)["cnt"]
    .mean()
    .reindex(range(1, 13))
)

month_names = [
    "Jan", "Feb", "Mar", "Apr", "May", "Jun",
    "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
]

plt.figure(figsize=(12, 6))
plt.plot(month_names, monthly_avg.values, marker="o")
plt.title("Average Daily Bike Rentals by Month (2011–2012)")
plt.xlabel("Month")
plt.ylabel("Average Daily Rentals")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()

# Rata-rata penyewaan berdasarkan musim
season_mapping = {
    1: "Winter",
    2: "Spring",
    3: "Summer",
    4: "Fall"
}

season_data = df_clean.copy()
season_data["season_name"] = season_data["season"].map(season_mapping)

season_avg = (
    season_data.groupby("season_name")["cnt"]
    .mean()
    .reindex(["Winter", "Spring", "Summer", "Fall"])
)

plt.figure(figsize=(9, 6))
sns.barplot(x=season_avg.index, y=season_avg.values)
plt.title("Average Daily Bike Rentals by Season (2011–2012)")
plt.xlabel("Season")
plt.ylabel("Average Daily Rentals")
plt.tight_layout()
plt.show()

print("Bulan dengan rata-rata tertinggi:",
      monthly_avg.idxmax(), "-", round(monthly_avg.max(), 2))
print("Bulan dengan rata-rata terendah:",
      monthly_avg.idxmin(), "-", round(monthly_avg.min(), 2))
print("Musim dengan rata-rata tertinggi:",
      season_avg.idxmax(), "-", round(season_avg.max(), 2))
print("Musim dengan rata-rata terendah:",
      season_avg.idxmin(), "-", round(season_avg.min(), 2))
```

### Bike Rentals Across Weather Conditions

```python
weather_mapping = {
    1: "Clear / Few Clouds",
    2: "Mist / Cloudy",
    3: "Light Rain / Snow",
    4: "Heavy Rain / Snow"
}

weather_data = df_clean.copy()
weather_data["weather_name"] = weather_data["weathersit"].map(
    weather_mapping
)

weather_avg = (
    weather_data.groupby("weather_name")["cnt"]
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(11, 6))
sns.barplot(x=weather_avg.index, y=weather_avg.values)

plt.title("Average Daily Bike Rentals by Weather Condition (2011–2012)")
plt.xlabel("Weather Condition")
plt.ylabel("Average Daily Rentals")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

print("Cuaca dengan rata-rata penyewaan tertinggi:",
      weather_avg.idxmax(), "-", round(weather_avg.max(), 2))
print("Cuaca dengan rata-rata penyewaan terendah:",
      weather_avg.idxmin(), "-", round(weather_avg.min(), 2))
```

**Insight:** (Opsional)
- Rata-rata penyewaan sepeda tertinggi terjadi pada kondisi cuaca Clear / Few Clouds sebesar 4.876,79 penyewaan per hari, sedangkan rata-rata terendah terjadi pada kondisi Light Rain / Snow sebesar 1.803,29 penyewaan per hari.



- Hasil analisis menunjukkan bahwa penyewaan sepeda cenderung lebih tinggi ketika cuaca cerah dibandingkan ketika terjadi hujan atau salju ringan. Hal ini mengindikasikan adanya hubungan antara kondisi cuaca dan permintaan penyewaan sepeda, meskipun faktor lain seperti musim, temperatur, dan hari kerja juga dapat memengaruhi jumlah penyewaan.

```python
# Mengelompokkan rata-rata penyewaan berdasarkan tahun
yearly_avg = (
    df_clean.groupby(df_clean["dteday"].dt.year)["cnt"]
    .mean()
)

display(yearly_avg.to_frame(name="Average Daily Rentals").round(2))

# Visualisasi perbandingan rata-rata penyewaan
plt.figure(figsize=(8, 5))
sns.barplot(x=yearly_avg.index.astype(str), y=yearly_avg.values)

plt.title("Average Daily Bike Rentals: 2011 vs 2012")
plt.xlabel("Year")
plt.ylabel("Average Daily Rentals")
plt.tight_layout()
plt.show()

# Menghitung persentase perubahan rata-rata
if 2011 in yearly_avg.index and 2012 in yearly_avg.index:
    percentage_change = (
        (yearly_avg.loc[2012] - yearly_avg.loc[2011])
        / yearly_avg.loc[2011] * 100
    )

    print(
        "Perubahan rata-rata penyewaan dari 2011 ke 2012:",
        round(percentage_change, 2), "%"
    )
```

**Insight:** (Opsional)
- Perbandingan rata-rata penyewaan harian antara 2011 dan 2012 digunakan untuk mengamati perubahan permintaan sepeda dari tahun ke tahun.



- Jika rata-rata tahun 2012 lebih tinggi daripada 2011, hal tersebut menunjukkan peningkatan penyewaan dalam periode pengamatan. Namun, data ini saja belum cukup untuk menjelaskan penyebab peningkatan tersebut.
